# Fleet Efficiency Benchmark — Analysis

Executed comparative analysis. The weighted MPG field is an analyst-created proxy, not an official EPA combined rating.

```python
from pathlib import Path
import json, pandas as pd
ROOT=Path('..').resolve()
df=pd.read_csv(ROOT/'data/processed/fleet_efficiency_clean.csv')
metrics=json.loads((ROOT/'reports/metrics.json').read_text())
```

```python
df.groupby('year').agg(observations=('model','size'),median_city=('cty','median'),median_highway=('hwy','median'),median_proxy=('combined_mpg_proxy','median')).round(2)
```

```python
df.groupby('class',observed=True).combined_mpg_proxy.median().sort_values(ascending=False).round(2)
```

```python
{'mann_whitney_u':metrics['mann_whitney_u'],'p_value':round(metrics['mann_whitney_pvalue'],3)}
```

The p-value does not support a clear 1999-vs-2008 distribution shift in this selected sample. Class and displacement provide more useful segmentation.
