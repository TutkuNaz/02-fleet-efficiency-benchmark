# Fleet Efficiency Benchmark — Paired Analysis

EPA combined MPG uses the harmonic 55% city / 45% highway consumption formula.

```python
from pathlib import Path
import json
import pandas as pd

ROOT = Path('..').resolve()
df = pd.read_csv(ROOT / 'data/processed/fleet_efficiency_clean.csv')
metrics = json.loads((ROOT / 'reports/metrics.json').read_text())
```

```python
model_year = (
    df.groupby(['manufacturer', 'model', 'year'], as_index=False)
      .agg(combined_mpg=('combined_mpg', 'median'))
)
paired = model_year.pivot(
    index=['manufacturer', 'model'], columns='year', values='combined_mpg'
).dropna()
paired.assign(change=paired[2008] - paired[1999]).describe().round(2)
```
