# Fleet Efficiency Benchmark — Data Quality

This notebook inspects the processed EPA-derived sample without changing the source observation balance.

```python
from pathlib import Path
import pandas as pd

ROOT = Path('..').resolve()
df = pd.read_csv(ROOT / 'data/processed/fleet_efficiency_clean.csv')
df.shape
```

```python
pd.DataFrame({
    'source_rows': [len(df)],
    'duplicate_source_rows': [int(df.duplicated().sum())],
    'null_cells': [int(df.isna().sum().sum())],
    'manufacturers': [df.manufacturer.nunique()],
    'models': [df.model.nunique()],
})
```

```python
df.groupby('year').agg(
    configurations=('model', 'size'),
    distinct_models=('model', 'nunique'),
).sort_index()
```

Exact duplicates are reported but retained. The inferential analysis aggregates configurations to model-year medians and pairs the same models across years.
